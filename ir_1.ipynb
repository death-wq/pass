```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

text = (
    "Information Retrieval systems are designed to retrieve relevant documents "
    "from large collections. Users are searching, browsing and ranking "
    "the retrieved results every day!"
)

# 1. Case folding and cleaning
clean = re.sub(r"[^a-zA-Z\\s]", " ", text.lower())

# 2. Tokenization
tokens = word_tokenize(clean)
print("Tokens         :", tokens)

# 3. Stop-word removal
stop_words = set(stopwords.words("english"))
filtered = [token for token in tokens if token not in stop_words]
print("After stop words:", filtered)

# 4. Stemming: compare Porter and Lancaster
porter = PorterStemmer()
lancaster = LancasterStemmer()

print(f"\\n{'Word':<14}{'Porter':<14}{'Lancaster':<14}")
for word in filtered:
    print(f"{word:<14}{porter.stem(word):<14}{lancaster.stem(word):<14}")

# 5. Statistics
print("\\nTokens before :", len(tokens))
print("Tokens after  :", len(filtered))
reduction = 100 * (len(tokens) - len(filtered)) / len(tokens) if tokens else 0
print("Reduction    : {:.1f}%".format(reduction))
print("Unique stems :", len({porter.stem(word) for word in filtered}))
```
